## PR: Anatomy
1. Outer segment: rhodopsin in discs
2. Inner segment: mitochondria + nucleus
3. Synaptic terminal
   - Synaptic ribbons with vesicles
   - Hundreds of synapses with horizontal cells and bipolar cells
   - Graded responses, not spikes
   - High glutamate release in the dark

Photoreceptors are "off devices":
- Light turns them off
- Light offset lets them depolarize again

Photoreceptors do not spike:
- They respond in a graded fashion to dark and light

## PR Membrane Potential Tracks Brightness
<figure style="float: right; width: 50%;">
  <img src="../assets/retina1_slide_phototransduction.jpg" alt="Description of the image" style="width: 100%; height: auto;">

- Resting potential: about -35 mV
- Very depolarized compared with typical neurons
- Light hyperpolarizes the photoreceptor

**Why is this useful?**
- In the gray zone, PRs release glutamate continuously
- Darkening / light offset: depolarization -> more glutamate
- Lightening / light onset: hyperpolarization -> less glutamate

# Photoreceptors, Bipolar Cells, and Horizontal Cells


## PR to Bipolar Cell (BC): Convergence and Divergence

<figure style="float: right; width: 66%;">
  <img src="../assets/2004kalat_fig56_convergenceInputOntoBpCells.png" alt="Description of the image" style="width: 100%; height: auto;">
  <figcaption style="text-align: center; font-size: 0.8em; color: gray;">
    Figure 5.6 from Kalat 2004, Biological Psychology.
  </figcaption>
</figure>

**Toward the retinal periphery:**
- More receptors converge onto bipolar and ganglion cells
- Summation helps detect faint lights
- But precision is lost: shape and exact location become less clear

**Key point:**
- We feel like we see the whole scene clearly
- In reality, high-resolution vision is mostly foveal
- The rest depends heavily on eye movements, context, and memory

## PR to BC: Synaptic Sign of Glutamate Effect

<figure style="float: right; width: 40%;">
  <img src="../assets/retina1_slide_pr2bc2gc.jpg" alt="Description of the image" style="width: 100%; height: auto;">
  <figcaption style="text-align: center; font-size: 0.8em; color: gray;">
  </figcaption>
</figure>

Unstimulated rods and cones steadily release glutamate.

Glutamate can have opposite effects on bipolar cells:
- **OFF bipolar cells:** sign-conserving synapse
  - Light off -> more glutamate -> depolarization
- **ON bipolar cells:** sign-inverting synapse
  - Light on -> less glutamate -> depolarization

**Odd but important:**
- Glutamate is usually excitatory
- Here, more glutamate hyperpolarizes ON bipolar cells
- Less glutamate depolarizes ON bipolar cells

<div style="clear: both;"></div>


## Horizontal Cells: Off Devices
<figure style="float: right; width: 55%;">
  <img src="../assets/retina1_slide_hcells_are_also_off_devices.jpg" alt="" style="width: 100%; height: auto;">
  <figcaption style="text-align: center; font-size: 0.8em; color: gray;">
  </figcaption>
</figure>

Retinal neurons often have concentric receptive fields:

1. ON-center / OFF-surround
2. OFF-center / ON-surround

_Note: the center and surround are always antagonistic._

<div style="clear: both;"></div>


## Horizontal Cells: Negative Feedback onto Cones
<figure style="float: right; width: 50%;">
  <img src="../assets/retina1_slide_hcells_negatively_feedback.jpg" alt="" style="width: 100%; height: auto;">
  <figcaption style="text-align: center; font-size: 0.8em; color: gray;">
  </figcaption>
</figure>

Horizontal cells send sign-inverting feedback onto cone terminals using GABA.

- If the horizontal cell is hyperpolarized: less inhibition
- Less inhibition means the cone terminal is disinhibited
- Disinhibition depolarizes the cone terminal

This process is called **lateral inhibition**.

One of the most important features of the visual system.

<div style="clear: both;"></div>


## Horizontal Cells: Opponent Surrounds
<figure style="float: right; width: 60%;">
  <img src="../assets/retina1_slide_hcells_responsible_for_opponent_surrounds.jpg" alt="" style="width: 100%; height: auto;">
  <figcaption style="text-align: center; font-size: 0.8em; color: gray;">
  </figcaption>
</figure>

Horizontal cells help build opponent surrounds.

- The center cone depolarizes to dark in its own location
- Surround cones respond through horizontal-cell feedback
- The cone terminal can therefore behave like an OFF-center / ON-surround unit

## PR + HC -> BC: The Center-Only Case
<figure style="float: right; width: 50%;">
  <img src="../assets/2004kalat_fig516_Pr2Bp_1.png"  style="width: 100%; height: auto;">
  <figcaption style="text-align: center; font-size: 0.8em; color: gray;">
    from Kalat 2004, Biological Psychology.
  </figcaption>
</figure>

**Center-only stimulation:**
- Light decreases photoreceptor output
- Less photoreceptor output changes bipolar-cell activity
- In the fovea, each cone connects to just one bipolar cell

**In this diagram:**
- Green arrows: excitation
- Red arrows: inhibition from horizontal cells
- Arrow width: strength of effect
- Lighter blue: more net excitation in a bipolar cell

**Key idea:**
- The center input can outweigh surround inhibition
- Nearby cells receive less direct excitation and more lateral inhibition

<div style="clear: both;"></div>


## PR + HC -> BC: The Center-Surround Case
<figure style="float: right; width: 50%;">
  <img src="../assets/2004kalat_fig516_Pr2Bp_2.png"  style="width: 100%; height: auto;">
  <figcaption style="text-align: center; font-size: 0.8em; color: gray;">
    from Kalat 2004, Biological Psychology.
  </figcaption>
</figure>

Now imagine light on receptors 6 through 10.

- Receptors 6-10 excite their bipolar cells
- They also activate the horizontal cell
- Bipolar cells 7, 8, and 9 receive inhibition from both sides
- Bipolar cells 6 and 10 receive inhibition from only one side
- Result: 6 and 10 respond the most

**What this illustrates:**
- Lateral inhibition reduces activity due to neighboring activity
- This heightens contrast
- Similar logic appears in touch, hearing, and olfaction

## From Bipolar Cells to RGCs

Bipolar cells also release glutamate.

For ganglion cells, glutamate is sign-preserving:
1. Light on -> ON bipolar cell depolarizes -> ON ganglion cell depolarizes
2. Light off -> OFF bipolar cell depolarizes -> OFF ganglion cell depolarizes

**Inheritance of center-surround response**

Ganglion cells inherit the center vs. surround receptive-field structure from bipolar-cell inputs.

<div style="clear: both;"></div>


## Response of OFF-Center Ganglion Cells
<figure style="float: right; width: 33%;">
  <img src="../chapter_illusions/retina1_slide_offcenter_rgc_firing.jpg" alt="Description of the image" style="width: 100%; height: auto;">
  <figcaption style="text-align: center; font-size: 0.8em; color: gray;">
  </figcaption>
</figure>

**OFF-center ganglion cell:**
- Dark spot in the OFF center -> higher firing
- Light annulus in the ON surround -> higher firing
- Uniform dark or light -> balanced excitation/inhibition -> weak response

**Best stimulus:**
- Black spot on white annulus
- OFF center excites
- ON surround excites

## Response of ON-Center Ganglion Cells

**ON-center ganglion cell:**
- Light spot in the ON center -> higher firing
- Dark annulus in the OFF surround -> higher firing
- Uniform dark or light -> balanced excitation/inhibition -> weak response

**Best stimulus:**
- White spot on black annulus
- ON center excites
- OFF surround excites

# Review

**Glutamate**
- Main excitatory transmitter in the CNS
- Rods and cones release glutamate
- Glutamate depolarizes OFF bipolar cells and horizontal cells
- ON and OFF bipolar cells release glutamate onto ganglion cells

**GABA**
- Main inhibitory transmitter in the CNS
- Horizontal cells release GABA back onto cone terminals
- This inhibits cone terminals

**The odd synapse**
- Cone -> ON bipolar cell
- More glutamate hyperpolarizes ON bipolar cells
- Less glutamate depolarizes ON bipolar cells

**Inheritance of center-surround responses**
- Bipolar-cell targets inherit antagonistic center vs. surround structure from cone inputs
- Ganglion-cell targets inherit center vs. surround structure from bipolar-cell inputs
