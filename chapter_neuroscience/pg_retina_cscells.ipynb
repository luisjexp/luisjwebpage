## What Makes CS Cells Special?

Like most visual neurons, CS cells have a **receptive field**.

**Location preference**
- A CS cell responds best to stimuli in one region of visual space
- The farther the stimulus is from that preferred location, the weaker the response

**Contrast polarity**
- Some CS cells prefer light spots surrounded by darkness
- Others prefer dark spots surrounded by light
- This is why they are called **center-surround** cells

**Size tuning**
- The best stimulus is not too large and not too small
- It should cover the center without spilling too far into the surround
- If the surround is also stimulated, the cell's response decreases

**Shape preference**
- A circular spot is usually the best stimulus
- A bar or strip can activate the cell, but usually less strongly
- Uniform stimulation across the whole receptive field gives a weak response

**Circular receptive fields**
- CS cells are not strongly tuned to orientation
- Compare this with orientation-tuned cells in V1

## Why Have Both ON and OFF Systems?

Ganglion cells have spontaneous firing rates of about 30 spikes/sec.

- They can increase to about 300 spikes/sec
- They can decrease only to 0 spikes/sec
- This gives more room to signal increases than decreases

**Why both systems?**
- Light edges and dark edges are both common in the world
- ON and OFF systems double the range of luminance changes that can be encoded
